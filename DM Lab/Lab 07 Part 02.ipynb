```python
# I.
import networkx as nx
import matplotlib.pyplot as plt

cities = ["A", "B", "C", "D", "E"]

G = nx.DiGraph()

G.add_edges_from([
    ("A", "B"),
    ("B", "C"),
    ("C", "D"),
    ("C", "E"),
    ("E", "D"),
    ("C", "B")
])

print("Adjacency List:")
print(nx.to_dict_of_lists(G))

print("Adjacency Matrix:")
print(nx.to_numpy_array(G, nodelist=cities, dtype=int))

# II.
nx.draw(G, with_labels=True, arrows=True)
plt.show()
```

```python
import networkx as nx

cities = ["A", "B", "C", "D", "E"]

G = nx.DiGraph()

G.add_edges_from([
    ("A", "B"),
    ("B", "C"),
    ("C", "D"),
    ("C", "E"),
    ("E", "D"),
    ("C", "B")
])

for city in cities:
    print(city)
    print("In-degree :", G.in_degree(city))
    print("Out-degree:", G.out_degree(city))
    print()

print("Self loops:")
print(list(nx.selfloop_edges(G)))
```

```python
# I.
import networkx as nx

G = nx.DiGraph()

G.add_edges_from([
    ("A", "B"),
    ("B", "C"),
    ("C", "D"),
    ("C", "E"),
    ("E", "D"),
    ("C", "B")
])

print("Is D reachable from A?")
if nx.has_path(G, "A", "D"):
    print("Yes, D is reachable from A")
else:
    print("No, D is not reachable from A")

print("All simple paths from A to D:")
print(list(nx.all_simple_paths(G, "A", "D")))


# II.
print("Cycles:")
print(list(nx.simple_cycles(G)))
```

```python
# I.
import networkx as nx

G = nx.DiGraph()

G.add_edges_from([
    (1, 2),
    (1, 3),
    (2, 4),
    (4, 2),
    (2, 5),
    (3, 5),
    (5, 3),
    (4, 6),
    (5, 6),
    (6, 7),
    (7, 8),
    (8, 6)
])

print("Cycles before removing:")
print(list(nx.simple_cycles(G)))

# II.
G.remove_edge(4, 2)
G.remove_edge(5, 3)
G.remove_edge(8, 6)

print("Cycles after removing:")
print(list(nx.simple_cycles(G)))

# III.
print("Shortest paths to 8:")

for node in range(1, 8):
    print(node, "to 8:", nx.shortest_path(G, node, 8))
```

```python
import networkx as nx

servers = ["S1", "S2", "S3", "S4", "S5"]

G = nx.DiGraph()

G.add_edges_from([
    ("S1", "S2"),
    ("S2", "S3"),
    ("S3", "S4"),
    ("S4", "S5"),
    ("S2", "S5")
])

for server in servers:
    print(server)
    print("Indegree :", G.in_degree(server))
    print("Outdegree:", G.out_degree(server))
    print()

print("Is S5 reachable from S1?")

if nx.has_path(G, "S1", "S5"):
    print("Yes, S5 is reachable from S1")
else:
    print("No, S5 is not reachable from S1")
```

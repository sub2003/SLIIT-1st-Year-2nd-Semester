```python
import networkx as nx
import matplotlib.pyplot as plt

M=nx.DiGraph()
M.add_nodes_from([1,2,3,4,5,6])
M.add_edges_from([(1,2),(2,3),(2,4),(1,5),(5,6)])
nx.draw(M,with_labels=True)
plt.show()
```

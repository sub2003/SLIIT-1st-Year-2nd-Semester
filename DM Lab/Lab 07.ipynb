```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Create a directed graph
G = nx.DiGraph()

# Add cities / vertices
cities = ["A", "B", "C", "D", "E"]
G.add_nodes_from(cities)

# Add flights / directed edges
G.add_edges_from([
    ("A", "B"),
    ("B", "C"),
    ("C", "D"),
    ("C", "E"),
    ("E", "D"),
    ("C", "B")
])

# -------- Adjacency List --------
print("Adjacency List:")

adj_list = nx.to_dict_of_lists(G)

for city in adj_list:
    print(city, ":", adj_list[city])


# -------- Adjacency Matrix --------
print("\nAdjacency Matrix:")

matrix = nx.to_numpy_array(G, nodelist=cities, dtype=int)

print("City Order:", cities)
print(matrix)


# -------- Display Graph --------
pos = {
    "A": (0, 2),
    "B": (2, 2),
    "C": (4, 2),
    "D": (6, 3),
    "E": (6, 1)
}

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=2000,
    arrows=True
)

#plt.title("Airline Flight Routes Directed Graph")
plt.show()
```

```python
print("Question 2: In-degree and Out-degree")

for city in cities:
    print(city)
    print("In-degree :", G.in_degree(city))
    print("Out-degree:", G.out_degree(city))
    print()

# Check self-loops
self_loops = list(nx.selfloop_edges(G))

print("Self-loops:", self_loops)

if len(self_loops) == 0:
    print("No self-loops available.")
else:
    print("Self-loops are available.")
```
